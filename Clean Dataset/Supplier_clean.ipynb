```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('Suppliers_v2.csv')
df.head()
```

```python
df.shape
```

```python
df.isnull().sum()    #4%
```

```python
df['rating'] = df['rating'].fillna(df.groupby('supplier_type')
    ['rating'].transform(lambda x:x.mode().iloc[0] if not x.mode().empty else np.nan))
```

```python
df.duplicated().sum()
```

```python
df.drop_duplicates(inplace=True)
```

```python
a = ['city','country','supplier_type']
for i in a:
    print(df[i].value_counts())
```

```python
df['city'] = df['city'].str.title().str.strip()
```

```python
df['city'].value_counts()
```

```python
df.drop(columns=['legacy_supplier_code'],inplace=True)
```

```python
df.head()
```

```python
a=df[['lead_time','rating']].describe(percentiles=[0.01,0.02,0.05,0.95,0.97,0.98,0.99]).T
a=a.iloc[:,3:]
a
```

```python
df['lead_time'] = np.where(df['lead_time']<df['lead_time'].quantile(0.01),
                           df['lead_time'].quantile(0.01),df['lead_time'])
```

```python
df['rating'] = np.where(df['rating']<0,0,df['rating'])
```

```python
df[['rating','lead_time']] = df[['rating','lead_time']].round(1)
```

```python
df.to_csv('Supplier.csv',index=False)
```
