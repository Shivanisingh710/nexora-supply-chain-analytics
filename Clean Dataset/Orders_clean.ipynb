### Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

### Read Dataset

```python
df = pd.read_csv('Orders_v2.csv')
df.head()
```

### Data Cleaning

#### Checking Missing Values

```python
df.isnull().sum()  #0.9% 
```

#### Handling Missing Values

```python
df['order_status'] = df['order_status'].str.strip().replace('', np.nan)
```

```python
df['order_status'].value_counts()
```

```python
a = ['Cancelled','Returned','Shipped']
df['order_status'] = df['order_status'].apply(lambda x: np.random.choice(a) if pd.isna(x) else x)
```

#### Detecting Extra Spaces

```python
x = ['order_status', 'shipment_status', 'payment_method', 'shipping_mode']
for i in x:
    print(df[i].unique())
```

```python
df = df.apply(lambda x: x.str.strip() if x.dtype == 'object' else x)
```

#### Checking Duplicates

```python
df.duplicated().sum()
```

#### Handling Duplicates

```python
df = df.drop_duplicates()
```

#### Checking Data Types

```python
df.dtypes
```

#### Converting Data Types

```python
df['order_date'] = pd.to_datetime(df['order_date'], format='mixed')
```

```python
df.dtypes
```

#### Removing Irrelevent Column

```python
df.drop(columns='session_id',inplace=True)
```

```python
df.head()
```

#### Saving Cleaned Dataset

```python
df.to_csv('Orders_Clean.csv', index=False)
```
