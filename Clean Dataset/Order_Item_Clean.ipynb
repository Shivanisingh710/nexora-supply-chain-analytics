#### Detecting Outliers

```python
cols = ['quantity','unit_price','discount']
a = df[cols].describe(percentiles=[0.01,0.02,0.05,0.95,0.97,0.98,0.99]).T
a = a.iloc[:,3:]
a
```

#### Handling Outliers

```python
print((df['discount']>40).sum())
print((df['unit_price'] < 0).sum())
print((df['quantity'] < 0).sum())
```

```python
a = ['quantity','unit_price']
for i in a:
    lower = df[i].quantile(0.01)
    df[i] = np.where(df[i]<lower,lower,df[i])
```

```python
df['discount'] = np.where(df['discount']>df['discount'].quantile(0.99),df['discount'].quantile(0.99),df['discount'])
```

```python
cols = ['quantity','unit_price','discount']
a = df[cols].describe(percentiles=[0.01,0.02,0.05,0.95,0.97,0.98,0.99]).T
a = a.iloc[:,3:]
a
```

#### Removing Irrelevent Column

```python
df.drop(columns='batch_reference',inplace=True)
```
